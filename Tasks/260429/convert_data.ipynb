```python
import pandas as pd
```

```python
df = pd.read_excel("../../Dataset/노이즈_제거_에지.xlsx")
df.head()
```

```python
result = df.groupby(["출처채널", "영상제목", "날짜", "링크"], as_index=False, sort=False).agg(
    Collab_Channel=("대상채널", lambda x: ", ".join(x.dropna().astype(str).unique()))
)

result = result.rename(columns={
    "출처채널": "Source_Channel",
    "영상제목": "Title",
    "날짜": "Date",
    "링크": "URL"
})
result = result[["Source_Channel", "Collab_Channel", "Title", "Date", "URL"]]
print(result.head(5))
```

```python
result = result.sort_values("Date", ascending=True)
result.to_csv("../../Dataset/data_260429.csv", index=False, encoding="utf-8-sig")

print(result.head(5))
```
